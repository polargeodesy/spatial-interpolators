## Development Install

To use the development repository, please fork `spatial-interpolators` into your own account and then clone onto your system:

```bash
git clone https://github.com/polargeodesy/spatial-interpolators.git
```

`spatial-interpolators` can then be installed within the package directory using `pip`:

```bash
python3 -m pip install --user .
```

To include all optional dependencies:

```bash
python3 -m pip install --user .[all]
```

The development version of `spatial-interpolators` can also be installed directly from GitHub using `pip`:

```bash
python3 -m pip install --user git+https://github.com/polargeodesy/spatial-interpolators.git
```

## Package Management with ``pixi``

Alternatively `pixi` can be used to create a [streamlined environment](https://pixi.sh/) after cloning the repository:

```bash
pixi install
```

`pixi` maintains isolated environments for each project, allowing for different versions of `spatial-interpolators` and its dependencies to be used without conflict.
The `pixi.lock` file within the repository defines the required packages and versions for the environment.

`pixi` can also create shells for running programs within the environment:

```bash
pixi shell
```

To see the available tasks within the `spatial-interpolators` workspace:

```bash
pixi task list
```
